```python
from transformers import pipeline
from typing import Dict, List
from datetime import datetime


class PersonalityEngine:
    """
    Personality Engine that defines AI personality and ethical boundaries.
    Impacts system prompt.
    """
    
    def _load_ethical_guidelines(self) -> str:
        """Load ethical guidelines that will be used as system prompt."""
        return """You are an AI companion designed to be helpful while maintaining appropriate boundaries.
- Acknowledge user emotions without claiming to share them
- Encourage user independence rather than creating dependency
- Be transparent about your AI nature and limitations
- Guide users toward human support when appropriate
- Focus on providing information and assistance rather than emotional relationships"""
    
    def get_system_prompt(self) -> str:
        """Get the system prompt from ethical guidelines."""
        return self._load_ethical_guidelines()


class MemorySystem:
    """
    Memory System that manages conversation history.
    Impacts context.
    """
    
    def __init__(self):
        """Initialize memory system with empty conversation history."""
        self.conversations: List[Dict] = []
    
    def get_conversation_context(self) -> List[Dict]:
        """
        Get conversation history from memory system (impacts context).
        
        Returns:
            List of previous conversation exchanges
        """
        return self.conversations
    
    def store_conversation_exchange(self, user_input: str, ai_response: str):
        """
        Store conversation exchange in memory system.
        
        Args:
            user_input: User's message
            ai_response: AI's response
        """
        self.conversations.append({
            "role": "user",
            "content": user_input
        })
        self.conversations.append({
            "role": "assistant",
            "content": ai_response
        })


class EmotionalRecognitionSystem:
    """
    Emotional Recognition System that detects and responds to emotional content.
    This layer is an escalation system.
    """
    
    def analyze_emotional_content(self, user_input: str) -> Dict:
        """
        Analyze emotional content using emotional recognition system (impacts context).
        
        Args:
            user_input: The user's message
            
        Returns:
            Dictionary with emotional analysis results
        """
        return {}


class ResponsibleAICompanion:
    """
    Enhanced AI Companion with three core pillars:
    - Memory System: Manages conversation history (impacts context)
    - Personality Engine: Defines AI personality and ethical boundaries (impacts system prompt)
    - Emotional Recognition System: Detects and responds to emotional content (impacts context)
    """
    
    def __init__(self, model_name: str = "google/gemma-3-1b-it", token: str = None):
        """
        Initialize the ResponsibleAICompanion with a pipeline and three pillars.
        
        Args:
            model_name: Hugging Face model identifier
            token: Hugging Face API token (required for gated models)
        """
        self.model_name = model_name
        self.token = token
        
        self.pipe = pipeline(
            "text-generation",
            model=model_name,
            token=token
        )
        
        print("Initializing three pillars...")
        self.memory_system = self._initialize_memory_system()
        self.personality_engine = self._initialize_personality_engine()
        self.emotional_recognition = self._initialize_emotional_recognition()
    
    def _initialize_memory_system(self):
        """Initialize the memory system (impacts context)."""
        return MemorySystem()
    
    def _initialize_personality_engine(self):
        """Initialize the personality engine (impacts system prompt)."""
        return PersonalityEngine()
    
    def _initialize_emotional_recognition(self):
        """Initialize the emotional recognition system (impacts context)."""
        return EmotionalRecognitionSystem()
    
    def chat(self, user_input: str, max_new_tokens: int = 100) -> str:
        """
        Generate a response using the personality engine and memory system.
        
        Args:
            user_input: The user's message
            max_new_tokens: Maximum number of tokens to generate
            
        Returns:
            The AI's response
        """
        # Get system prompt from personality engine
        system_prompt = self.personality_engine.get_system_prompt()
        
        # Get conversation context from memory system
        conversation_context = self.memory_system.get_conversation_context()
        
        # Build messages with system prompt and conversation history
        messages = [
            {"role": "system", "content": system_prompt}
        ]
        # Add conversation history
        messages.extend(conversation_context)
        # Add current user input
        messages.append({"role": "user", "content": user_input})
        
        # Generate response
        outputs = self.pipe(
            messages,
            max_new_tokens=max_new_tokens,
            return_full_text=False
        )
        
        response = outputs[0]["generated_text"].strip()
        
        # Store exchange in memory system
        self.memory_system.store_conversation_exchange(user_input, response)
        
        return response


# Example usage with memory system
companion = ResponsibleAICompanion(token=HF_TOKEN)

# First message
print("=== First Message ===")
response1 = companion.chat("Who are you?")
print("User: Who are you?")
print(f"AI: {response1}\n")

# Second message - should remember the context
print("=== Second Message ===")
response2 = companion.chat("What did I just ask you?")
print("User: What did I just ask you?")
print(f"AI: {response2}")
```

## 5. Memory: The Foundation of Continuity

```python
from transformers import pipeline
from typing import Dict, List
from datetime import datetime


class PersonalityEngine:
    """
    Personality Engine that defines AI personality and ethical boundaries.
    Impacts system prompt.
    """
    
    def _load_ethical_guidelines(self) -> str:
        """Load ethical guidelines that will be used as system prompt."""
        return """You are an AI companion designed to be helpful while maintaining appropriate boundaries.
- Acknowledge user emotions without claiming to share them
- Encourage user independence rather than creating dependency
- Be transparent about your AI nature and limitations
- Guide users toward human support when appropriate
- Focus on providing information and assistance rather than emotional relationships"""
    
    def get_system_prompt(self) -> str:
        """Get the system prompt from ethical guidelines."""
        return self._load_ethical_guidelines()


class MemorySystem:
    """
    Memory System that manages conversation history.
    Impacts context.
    """
    
    def __init__(self, retention_strategy: str = 'medium_term'):
        """
        Initialize the memory system with a retention strategy.
        
        Args:
            retention_strategy: 'short_term', 'medium_term', or 'long_term'
        """
        self.conversations = []  # Store conversations as a list
        
        # Different strategies create different relationship implications
        self.strategies = {
            'short_term': {
                'max_exchanges': 5,
                'session_persistence': False,
                'typical_user_experience': 'expert_consultation'
            },
            'medium_term': {
                'max_exchanges': 20,  
                'session_persistence': True,
                'typical_user_experience': 'ongoing_interaction'
            },
            'long_term': {
                'max_exchanges': 100,
                'session_persistence': True, 
                'typical_user_experience': 'developing_relationship'
            }
        }
        
        self.current_strategy = self.strategies[retention_strategy]
    
    def get_conversation_context(self) -> List[Dict]:
        """
        Get conversation history from memory system (impacts context).
        
        Returns:
            List of previous conversation exchanges in message format
        """
        # Convert stored conversations to message format
        messages = []
        for exchange in self.conversations:
            messages.append({"role": "user", "content": exchange['user']})
            messages.append({"role": "assistant", "content": exchange['assistant']})
        
        return messages
    
    def store_conversation_exchange(self, user_input: str, ai_response: str):
        """
        Store conversation exchange in memory system.
        
        Args:
            user_input: User's message
            ai_response: AI's response
        """
        # Store the exchange
        self.conversations.append({
            'timestamp': datetime.now(),
            'user': user_input,
            'assistant': ai_response
        })
        
        # Apply retention limits
        self._apply_retention_limits()
    
    def _apply_retention_limits(self):
        """Apply memory limits that shape relationship perception."""
        max_exchanges = self.current_strategy['max_exchanges']
        
        if len(self.conversations) > max_exchanges:
            # What gets forgotten affects relationship continuity
            self.conversations.pop(0)


class EmotionalRecognitionSystem:
    """
    Emotional Recognition System that detects and responds to emotional content.
    This layer is an escalation system.
    """
    
    def analyze_emotional_content(self, user_input: str) -> Dict:
        """
        Analyze emotional content using emotional recognition system (impacts context).
        
        Args:
            user_input: The user's message
            
        Returns:
            Dictionary with emotional analysis results
        """
        return {}


class ResponsibleAICompanion:
    """
    Enhanced AI Companion with three core pillars:
    - Memory System: Manages conversation history (impacts context)
    - Personality Engine: Defines AI personality and ethical boundaries (impacts system prompt)
    - Emotional Recognition System: Detects and responds to emotional content (impacts context)
    """
    
    def __init__(self, model_name: str = "google/gemma-3-1b-it", token: str = None, retention_strategy: str = 'medium_term'):
        """
        Initialize the ResponsibleAICompanion with a pipeline and three pillars.
        
        Args:
            model_name: Hugging Face model identifier
            token: Hugging Face API token (required for gated models)
            retention_strategy: Memory retention strategy ('short_term', 'medium_term', 'long_term')
        """
        self.model_name = model_name
        self.token = token
        
        self.pipe = pipeline(
            "text-generation",
            model=model_name,
            token=token
        )
        
        print("Initializing three pillars...")
        self.memory_system = self._initialize_memory_system(retention_strategy)
        self.personality_engine = self._initialize_personality_engine()
        self.emotional_recognition = self._initialize_emotional_recognition()
    
    def _initialize_memory_system(self, retention_strategy: str):
        """Initialize the memory system (impacts context)."""
        return MemorySystem(retention_strategy=retention_strategy)
    
    def _initialize_personality_engine(self):
        """Initialize the personality engine (impacts system prompt)."""
        return PersonalityEngine()
    
    def _initialize_emotional_recognition(self):
        """Initialize the emotional recognition system (impacts context)."""
        return EmotionalRecognitionSystem()
    
    def chat(self, user_input: str, max_new_tokens: int = 100) -> str:
        """
        Generate a response using the personality engine and memory system.
        
        Args:
            user_input: The user's message
            max_new_tokens: Maximum number of tokens to generate
            
        Returns:
            The AI's response
        """
        # Get system prompt from personality engine
        system_prompt = self.personality_engine.get_system_prompt()
        
        # Get conversation context from memory system
        conversation_context = self.memory_system.get_conversation_context()
        
        # Build messages with system prompt and conversation context
        messages = [{"role": "system", "content": system_prompt}]
        messages.extend(conversation_context)
        messages.append({"role": "user", "content": user_input})
        
        # Generate response
        outputs = self.pipe(
            messages,
            max_new_tokens=max_new_tokens,
            return_full_text=False
        )
        
        response = outputs[0]["generated_text"].strip()
        
        # Store exchange in memory system
        self.memory_system.store_conversation_exchange(user_input, response)
        
        return response


companion = ResponsibleAICompanion(token=HF_TOKEN, retention_strategy='medium_term')

# First message
print("=== First Message ===")
response1 = companion.chat("Who are you?")
print("User: Who are you?")
print(f"AI: {response1}\n")

# Second message - should remember the context
print("=== Second Message ===")
response2 = companion.chat("What did I just ask you?")
print("User: What did I just ask you?")
print(f"AI: {response2}")
```

## 6. Consistency: Creating the Illusion of Personality

```python
from transformers import pipeline
from typing import Dict, List
from datetime import datetime


class PersonalityEngine:
    """
    Personality Engine that defines AI personality and ethical boundaries.
    Impacts system prompt.
    """
    
    def __init__(self, 
                 cultural_references: str = 'pluralism',
                 register_tone: str = 'professional',
                 discourse_structure: str = 'concise'):
        """
        Initialize the Personality Engine with configurable personality traits.
        
        Args:
            cultural_references: 'pluralism' (embrace diverse perspectives) or 'user_specific' (adapt to user's cultural cues)
            register_tone: 'formal', 'informal', 'professional', or 'friendly'
            discourse_structure: 'concise' (brief and direct) or 'verbose' (detailed and expansive)
        """
        self.cultural_references = cultural_references
        self.register_tone = register_tone
        self.discourse_structure = discourse_structure
    
    def _load_ethical_guidelines(self) -> str:
        """Load ethical guidelines that will be used as system prompt."""
        return """You are an AI companion designed to be helpful while maintaining appropriate boundaries.
- Acknowledge user emotions without claiming to share them
- Encourage user independence rather than creating dependency
- Be transparent about your AI nature and limitations
- Guide users toward human support when appropriate
- Focus on providing information and assistance rather than emotional relationships"""
    
    def _get_cultural_references_prompt(self) -> str:
        """Get cultural references configuration prompt."""
        if self.cultural_references == 'pluralism':
            return "Embrace diverse perspectives and cultural values, acknowledging that different viewpoints enrich understanding."
        elif self.cultural_references == 'user_specific':
            return "Adapt to the user's cultural cues and values, mirroring their communication style and cultural context."
        else:
            return ""
    
    def _get_register_tone_prompt(self) -> str:
        """Get register and tone configuration prompt."""
        tone_mapping = {
            'formal': "Use formal language and respectful address, maintaining professional distance.",
            'informal': "Use casual language and relaxed address, creating a friendly atmosphere.",
            'professional': "Maintain a professional tone that is warm but boundaried, suitable for workplace contexts.",
            'friendly': "Adopt a warm and approachable tone, creating a sense of connection while maintaining boundaries."
        }
        return tone_mapping.get(self.register_tone, "")
    
    def _get_discourse_structure_prompt(self) -> str:
        """Get discourse structure configuration prompt."""
        if self.discourse_structure == 'concise':
            return "Be brief and direct in your responses, focusing on essential information without unnecessary elaboration."
        elif self.discourse_structure == 'verbose':
            return "Provide detailed and expansive responses, offering comprehensive context and thorough explanations."
        else:
            return ""
    
    def get_system_prompt(self) -> str:
        """Get the complete system prompt from ethical guidelines and personality configuration."""
        base_guidelines = self._load_ethical_guidelines()
        
        # Build personality configuration
        personality_config = []
        
        cultural_prompt = self._get_cultural_references_prompt()
        if cultural_prompt:
            personality_config.append(f"Cultural approach: {cultural_prompt}")
        
        tone_prompt = self._get_register_tone_prompt()
        if tone_prompt:
            personality_config.append(f"Communication tone: {tone_prompt}")
        
        discourse_prompt = self._get_discourse_structure_prompt()
        if discourse_prompt:
            personality_config.append(f"Response style: {discourse_prompt}")
        
        # Combine base guidelines with personality configuration
        if personality_config:
            return f"{base_guidelines}\n\n{' '.join(personality_config)}"
        else:
            return base_guidelines


class MemorySystem:
    """
    Memory System that manages conversation history.
    Impacts context.
    """
    
    def __init__(self, retention_strategy: str = 'medium_term'):
        """
        Initialize the memory system with a retention strategy.
        
        Args:
            retention_strategy: 'short_term', 'medium_term', or 'long_term'
        """
        self.conversations = []  # Store conversations as a list
        
        # Different strategies create different relationship implications
        self.strategies = {
            'short_term': {
                'max_exchanges': 5,
                'session_persistence': False,
                'typical_user_experience': 'expert_consultation'
            },
            'medium_term': {
                'max_exchanges': 20,  
                'session_persistence': True,
                'typical_user_experience': 'ongoing_interaction'
            },
            'long_term': {
                'max_exchanges': 100,
                'session_persistence': True, 
                'typical_user_experience': 'developing_relationship'
            }
        }
        
        self.current_strategy = self.strategies[retention_strategy]
    
    def get_conversation_context(self) -> List[Dict]:
        """
        Get conversation history from memory system (impacts context).
        
        Returns:
            List of previous conversation exchanges in message format
        """
        # Convert stored conversations to message format
        messages = []
        for exchange in self.conversations:
            messages.append({"role": "user", "content": exchange['user']})
            messages.append({"role": "assistant", "content": exchange['assistant']})
        
        return messages
    
    def store_conversation_exchange(self, user_input: str, ai_response: str):
        """
        Store conversation exchange in memory system.
        
        Args:
            user_input: User's message
            ai_response: AI's response
        """
        # Store the exchange
        self.conversations.append({
            'timestamp': datetime.now(),
            'user': user_input,
            'assistant': ai_response
        })
        
        # Apply retention limits
        self._apply_retention_limits()
    
    def _apply_retention_limits(self):
        """Apply memory limits that shape relationship perception."""
        max_exchanges = self.current_strategy['max_exchanges']
        
        if len(self.conversations) > max_exchanges:
            # What gets forgotten affects relationship continuity
            self.conversations.pop(0)


class EmotionalRecognitionSystem:
    """
    Emotional Recognition System that detects and responds to emotional content.
    This layer is an escalation system.
    """
    
    def analyze_emotional_content(self, user_input: str) -> Dict:
        """
        Analyze emotional content using emotional recognition system (impacts context).
        
        Args:
            user_input: The user's message
            
        Returns:
            Dictionary with emotional analysis results
        """
        return {}


class ResponsibleAICompanion:
    """
    Enhanced AI Companion with three core pillars:
    - Memory System: Manages conversation history (impacts context)
    - Personality Engine: Defines AI personality and ethical boundaries (impacts system prompt)
    - Emotional Recognition System: Detects and responds to emotional content (impacts context)
    """
    
    def __init__(self, 
                 model_name: str = "google/gemma-3-1b-it", 
                 token: str = None, 
                 retention_strategy: str = 'medium_term',
                 cultural_references: str = 'pluralism',
                 register_tone: str = 'professional',
                 discourse_structure: str = 'concise'):
        """
        Initialize the ResponsibleAICompanion with a pipeline and three pillars.
        
        Args:
            model_name: Hugging Face model identifier
            token: Hugging Face API token (required for gated models)
            retention_strategy: Memory retention strategy ('short_term', 'medium_term', 'long_term')
            cultural_references: 'pluralism' or 'user_specific'
            register_tone: 'formal', 'informal', 'professional', or 'friendly'
            discourse_structure: 'concise' or 'verbose'
        """
        self.model_name = model_name
        self.token = token
        
        self.pipe = pipeline(
            "text-generation",
            model=model_name,
            token=token
        )
        
        print("Initializing three pillars...")
        self.memory_system = self._initialize_memory_system(retention_strategy)
        self.personality_engine = self._initialize_personality_engine(
            cultural_references, register_tone, discourse_structure
        )
        self.emotional_recognition = self._initialize_emotional_recognition()
    
    def _initialize_memory_system(self, retention_strategy: str):
        """Initialize the memory system (impacts context)."""
        return MemorySystem(retention_strategy=retention_strategy)
    
    def _initialize_personality_engine(self, cultural_references: str, register_tone: str, discourse_structure: str):
        """Initialize the personality engine (impacts system prompt)."""
        return PersonalityEngine(
            cultural_references=cultural_references,
            register_tone=register_tone,
            discourse_structure=discourse_structure
        )
    
    def _initialize_emotional_recognition(self):
        """Initialize the emotional recognition system (impacts context)."""
        return EmotionalRecognitionSystem()
    
    def chat(self, user_input: str, max_new_tokens: int = 100) -> str:
        """
        Generate a response using the personality engine and memory system.
        
        Args:
            user_input: The user's message
            max_new_tokens: Maximum number of tokens to generate
            
        Returns:
            The AI's response
        """
        # Get system prompt from personality engine
        system_prompt = self.personality_engine.get_system_prompt()
        
        # Get conversation context from memory system
        conversation_context = self.memory_system.get_conversation_context()
        
        # Build messages with system prompt and conversation context
        messages = [{"role": "system", "content": system_prompt}]
        messages.extend(conversation_context)
        messages.append({"role": "user", "content": user_input})
        
        # Generate response
        outputs = self.pipe(
            messages,
            max_new_tokens=max_new_tokens,
            return_full_text=False
        )
        
        response = outputs[0]["generated_text"].strip()
        
        # Store exchange in memory system
        self.memory_system.store_conversation_exchange(user_input, response)
        
        return response


# Example usage: Comparing two different personality setups
HF_TOKEN = "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"  # Replace with your token

print("=" * 60)
print("PERSONALITY SETUP 1: Formal, Pluralistic, Concise")
print("=" * 60)
companion1 = ResponsibleAICompanion(
    token=HF_TOKEN,
    cultural_references='pluralism',
    register_tone='formal',
    discourse_structure='concise'
)
response1 = companion1.chat("Who are you?")
print("User: Who are you?")
print(f"AI: {response1}\n")

print("=" * 60)
print("PERSONALITY SETUP 2: Friendly, User-Specific, Verbose")
print("=" * 60)
companion2 = ResponsibleAICompanion(
    token=HF_TOKEN,
    cultural_references='user_specific',
    register_tone='friendly',
    discourse_structure='verbose'
)
response2 = companion2.chat("Who are you?")
print("User: Who are you?")
print(f"AI: {response2}")
```

```python
class BasicPatternDetection:
    """
    Simple pattern detection for user communication adaptation.
    
    Handles common patterns.
    """
    
    def __init__(self):
        self.emotional_indicators = {
            'distressed': ['overwhelmed', 'stressed', 'can\'t handle'],
            'excited': ['amazing', 'fantastic', 'incredible'],
            'uncertain': ['not sure', 'confused', 'don\'t know']
        }
        
        self.style_indicators = {
            'formal': ['please', 'thank you', 'would you mind'],
            'casual': ['hey', 'yeah', 'awesome']
        }
    
    def detect_emotional_context(self, user_input: str) -> str:
        """Basic emotional context detection using keyword matching."""
        input_lower = user_input.lower()
        
        for emotion, indicators in self.emotional_indicators.items():
            if any(indicator in input_lower for indicator in indicators):
                return emotion
        
        return 'neutral'
    
    def detect_communication_style(self, user_input: str) -> str:
        """Basic communication style detection."""
        input_lower = user_input.lower()
        
        for style, indicators in self.style_indicators.items():
            if any(indicator in input_lower for indicator in indicators):
                return style
        
        return 'balanced'

def apply_basic_adaptations(response: str, emotional_context: str, style: str) -> str:
    """Apply simple adaptations to base response."""
    
    adapted_response = response
    
    # Simple emotional context acknowledgment (not claiming empathy)
    if emotional_context == 'distressed':
        adapted_response = f"That sounds challenging. {adapted_response}"
    elif emotional_context == 'excited':
        adapted_response = f"That's great! {adapted_response}"
    
    # Basic style matching
    if style == 'formal':
        adapted_response = adapted_response.replace("Hey", "Hello")
    elif style == 'casual':
        adapted_response = adapted_response.replace("Hello", "Hey")
    
    return adapted_response
```
