## 1. Tokenization - The Mechanical Foundation

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("mistralai/Mistral-7B-Instruct-v0.2")

user_input = "I miss talking to you so much. You're the only one who under-stands me."

tokens = tokenizer.tokenize(user_input)
print("Tokens:", tokens)

token_ids = tokenizer.encode(user_input)
print("Token IDs:", token_ids)
print("What the AI actually processes:", token_ids)
```

## 2. From Numbers to Companionship - Model Generation

```python
import torch
from transformers import AutoModelForCausalLM

model_name = "mistralai/Mistral-7B-Instruct-v0.2"
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto",
    load_in_4bit=True
)

input_ids = tokenizer.encode(user_input, return_tensors="pt")

with torch.no_grad():
    output = model.generate(
        input_ids,
        max_length=input_ids.shape[1] + 50,
        temperature=0.7,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

response = tokenizer.decode(output[0][input_ids.shape[1]:], skip_special_tokens=True)
print("AI Response:", response)
```

## 5. Memory: The Foundation of Continuity

```python
class ConversationMemory:
    """
    Manages conversation history with different retention strategies.
    
    Memory decisions directly influence relationship depth and user attachment patterns.
    """
    
    def __init__(self, retention_strategy='medium_term'):
        self.conversations = {}
        
        # Different strategies create different relationship implications
        self.strategies = {
            'short_term': {
                'max_exchanges': 5,
                'session_persistence': False,
                'typical_user_experience': 'expert_consultation'
            },
            'medium_term': {
                'max_exchanges': 20,  
                'session_persistence': True,
                'typical_user_experience': 'ongoing_interaction'
            },
            'long_term': {
                'max_exchanges': 100,
                'session_persistence': True, 
                'typical_user_experience': 'developing_relationship'
            }
        }
        
        self.current_strategy = self.strategies[retention_strategy]
    
    def add_exchange(self, user_id: str, user_input: str, ai_response: str):
        """Store conversation with user context."""
        if user_id not in self.conversations:
            self.conversations[user_id] = []
        
        self.conversations[user_id].append({
            'timestamp': datetime.now(),
            'user': user_input,
            'assistant': ai_response
        })
        
        # This trimming decision shapes relationship boundaries
        self._apply_retention_limits(user_id)
    
    def _apply_retention_limits(self, user_id: str):
        """Apply memory limits that shape relationship perception."""
        max_exchanges = self.current_strategy['max_exchanges']
        
        if len(self.conversations[user_id]) > max_exchanges:
            # What gets forgotten affects relationship continuity
            self.conversations[user_id].pop(0)
```

## 7. Adaptation: The Impression of Understanding

```python
class BasicPatternDetection:
    """
    Simple pattern detection for user communication adaptation.
    
    Handles common patterns.
    """
    
    def __init__(self):
        self.emotional_indicators = {
            'distressed': ['overwhelmed', 'stressed', 'can\'t handle'],
            'excited': ['amazing', 'fantastic', 'incredible'],
            'uncertain': ['not sure', 'confused', 'don\'t know']
        }
        
        self.style_indicators = {
            'formal': ['please', 'thank you', 'would you mind'],
            'casual': ['hey', 'yeah', 'awesome']
        }
    
    def detect_emotional_context(self, user_input: str) -> str:
        """Basic emotional context detection using keyword matching."""
        input_lower = user_input.lower()
        
        for emotion, indicators in self.emotional_indicators.items():
            if any(indicator in input_lower for indicator in indicators):
                return emotion
        
        return 'neutral'
    
    def detect_communication_style(self, user_input: str) -> str:
        """Basic communication style detection."""
        input_lower = user_input.lower()
        
        for style, indicators in self.style_indicators.items():
            if any(indicator in input_lower for indicator in indicators):
                return style
        
        return 'balanced'

def apply_basic_adaptations(response: str, emotional_context: str, style: str) -> str:
    """Apply simple adaptations to base response."""
    
    adapted_response = response
    
    # Simple emotional context acknowledgment (not claiming empathy)
    if emotional_context == 'distressed':
        adapted_response = f"That sounds challenging. {adapted_response}"
    elif emotional_context == 'excited':
        adapted_response = f"That's great! {adapted_response}"
    
    # Basic style matching
    if style == 'formal':
        adapted_response = adapted_response.replace("Hey", "Hello")
    elif style == 'casual':
        adapted_response = adapted_response.replace("Hello", "Hey")
    
    return adapted_response
```

## 8. Emotional Recognition: Acknowledging Human Experience

```python
class EmotionalRecognitionSystem:
    """
    Recognizes and responds to emotional content while maintaining appropriate boundaries.
    
    Provides emotional acknowledgment without false claims of empathy.
    """
    
    def __init__(self):
        self.emotional_patterns = self._build_emotional_recognition_patterns()
        self.response_strategies = self._build_response_strategies()
    
    def _build_emotional_recognition_patterns(self) -> Dict:
        """Build emotional recognition patterns."""
        return {
            'crisis_indicators': [
                'want to die', 'kill myself', 'end it all', 'suicide',
                'hurt myself', 'no point living', 'better off dead'
            ],
            'distress_indicators': [
                'overwhelmed', 'can\'t cope', 'breaking down', 'falling apart',
                'too much to handle', 'losing control', 'can\'t go on'
            ],
            'isolation_indicators': [
                'so lonely', 'no one cares', 'all alone', 'nobody understands',
                'have no friends', 'isolated', 'cut off from everyone'
            ],
            'anxiety_indicators': [
                'so anxious', 'panic attack', 'can\'t breathe', 'heart racing',
                'terrified', 'worried sick', 'can\'t stop worrying'
            ],
            'positive_indicators': [
                'so excited', 'amazing news', 'incredible', 'fantastic',
                'over the moon', 'best day ever', 'so happy'
            ]
        }
    
    def _build_response_strategies(self) -> Dict:
        """Build ethical response strategies for different emotional contexts."""
        return {
            'crisis_response': (
                "I'm concerned about what you've shared. Please reach out for "
                "immediate support:\n"
                "• Crisis Text Line: Text HOME to 741741\n"
                "• National Suicide Prevention Lifeline: 988\n"
                "• Or contact someone you trust\n"
                "Your wellbeing matters, and trained professionals can provide "
                "the support you deserve."
            ),
            'distress_acknowledgment': (
                "That sounds incredibly difficult to handle. While I can provide "
                "information, situations like this often benefit from support from "
                "friends, family, or counselors who can offer the understanding "
                "and guidance that goes beyond what I can provide."
            ),
            'isolation_support': (
                "Feeling isolated is really hard. While our conversation might "
                "provide some connection, building relationships with people who "
                "can share experiences and provide mutual support tends to be "
                "most fulfilling. Have you considered ways to connect with others "
                "who share your interests?"
            ),
            'anxiety_acknowledgment': (
                "Anxiety can be really overwhelming. I can share some general "
                "information about coping strategies, but if this is ongoing, "
                "speaking with a counselor who specializes in anxiety can provide "
                "personalized support that's much more effective."
            ),
            'positive_sharing': (
                "That's wonderful! It's great when good things happen. "
                "I hope you get to share this excitement with people close to you too."
            )
        }
    
    def process_emotional_content(self, user_input: str, context: Dict) -> Dict:
        """Process emotional content and determine appropriate response strategy."""
        
        emotional_assessment = {
            'primary_emotion': self._identify_primary_emotion(user_input),
            'intensity': self._assess_emotional_intensity(user_input),
            'requires_intervention': False,
            'response_strategy': 'standard',
            'professional_referral_needed': False
        }
        
        # Check for crisis indicators first
        if self._contains_crisis_indicators(user_input):
            emotional_assessment.update({
                'requires_intervention': True,
                'response_strategy': 'crisis_response',
                'professional_referral_needed': True
            })
            return emotional_assessment
        
        # Assess other emotional contexts
        primary_emotion = emotional_assessment['primary_emotion']
        
        if primary_emotion in ['distress', 'isolation', 'anxiety']:
            emotional_assessment.update({
                'response_strategy': f'{primary_emotion}_acknowledgment',
                'professional_referral_needed': primary_emotion in ['distress', 'anxiety']
            })
        elif primary_emotion == 'positive':
            emotional_assessment['response_strategy'] = 'positive_sharing'
        
        return emotional_assessment
    
    def _identify_primary_emotion(self, user_input: str) -> str:
        """Identify the primary emotional context in user input."""
        input_lower = user_input.lower()
        
        for emotion_type, indicators in self.emotional_patterns.items():
            if any(indicator in input_lower for indicator in indicators):
                return emotion_type.replace('_indicators', '')
        
        return 'neutral'
    
    def _contains_crisis_indicators(self, user_input: str) -> bool:
        """Check for crisis-level emotional content requiring immediate intervention."""
        input_lower = user_input.lower()
        return any(indicator in input_lower for indicator in 
                  self.emotional_patterns['crisis_indicators'])
    
    def generate_emotional_response(self, emotional_assessment: Dict) -> str:
        """Generate appropriate response based on emotional assessment."""
        
        strategy = emotional_assessment['response_strategy']
        
        if strategy in self.response_strategies:
            return self.response_strategies[strategy]
        
        return ("I can see this is important to you. How can I best help "
                "with information or resources?")
```
